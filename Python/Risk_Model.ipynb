```python
import numpy as np
```

```python
import pandas as pd

#data processing

patients = pd.read_csv(r"C:\Users\DELL\OneDrive\Desktop\Health Outcome Analysis & Risk Stratification\Excel_File\Patients(1).csv")
diagnoses = pd.read_csv(r"C:\Users\DELL\OneDrive\Desktop\Health Outcome Analysis & Risk Stratification\Excel_File\diagnoses.csv")
outcome = pd.read_csv(r"C:\Users\DELL\OneDrive\Desktop\Health Outcome Analysis & Risk Stratification\Excel_File\outcomes.csv")
labs = pd.read_csv(r"C:\Users\DELL\OneDrive\Desktop\Health Outcome Analysis & Risk Stratification\Excel_File\labs(1).csv")
```

```python
patients = patients.merge(diagnoses, on = 'DiagnosisID')
patients = patients.merge(outcome, on = 'OutcomeID')
```

```python
patients['AdmissionDate'] = pd.to_datetime(
    patients['AdmissionDate'],
    dayfirst=True
)

patients['DischargeDate'] = pd.to_datetime(
    patients['DischargeDate'],
    dayfirst=True
)

patients['LengthOfStay'] = (
    patients['DischargeDate'] - patients['AdmissionDate']
).dt.days
```

```python
patients['OutcomeEncoded'] = patients['OutcomeName_x'].map({
    'Recovered': 0,
    'Complicated': 1,
    'Deceased': 1
})
```

```python
patients['HighRisk'] = np.where((patients['Age']>65) | (patients['OutcomeName_x'].isin(['complicated', 'Deceased'])), 1,0)
```

```python
abnormal_conditions = { 
  'Blood Sugar' : lambda x: x>120, 
  'Cholestrol' : lambda x: x>200, 
  'Hemoglobin' : lambda x: x<13 
} 

def count_abnormal_labs(patient_id): 
    patient_labs = labs[labs['PatientID']== patient_id] 
    count = 0 
    
    for test_name, condition in abnormal_conditions.items(): 
        test_results = patient_labs[patient_labs['TestName'] == test_name] 
        count += test_results['Result'].apply(condition).sum() 
    return count 

patients['AbnormalLabCount'] = patients['PatientID'].apply(count_abnormal_labs)
```

```python
features = patients[['Age', 'LengthOfstay', 'TreatmentCost']]
target = patients['OutcomeEncoded']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.3,
    random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    solver="lbfgs",
    max_iter=500,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred = model.predict(X_test)

print("Classification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

# Probability predictions
y_pred = model.predict_proba(X_test)[:, 1]

# ROC Curve
fpr, tpr, thresholds = roc_curve(y_test, y_pred)

# AUC Score
roc_auc = auc(fpr, tpr)

# Plot
plt.figure()
plt.plot(
    fpr,
    tpr,
    color='blue',
    label=f'ROC Curve (AUC = {roc_auc:.2f})'
)

plt.plot([0, 1], [0, 1],
         color='gray',
         linestyle='--')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC)')
plt.legend(loc='lower right')

plt.show()
```
